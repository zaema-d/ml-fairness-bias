```python
import sys
print(sys.executable)  # sanity check

import pandas as pd
from aif360.datasets import GermanDataset
from ucimlrepo import fetch_ucirepo
from folktables import ACSDataSource, ACSEmployment

pd.set_option('display.max_columns', None)
```

German Credit Dataset

```python
german = GermanDataset()
german_df, _ = german.convert_to_dataframe()

print("=== GERMAN CREDIT ===")
print("Shape:", german_df.shape)
print("Columns:", list(german_df.columns))
print("Protected attributes:", german.protected_attribute_names)
german_df.head()
```

Taiwan Credit Default

```python
taiwan = fetch_ucirepo(id=350)
taiwan_df = taiwan.data.features.copy()
taiwan_df['default'] = taiwan.data.targets

print("=== TAIWAN CREDIT DEFAULT ===")
print("Shape:", taiwan_df.shape)
print("Columns:", list(taiwan_df.columns))
taiwan_df.head()
```

```python
# Taiwan — check actual column names
taiwan_raw = fetch_ucirepo(id=350)
df_check = taiwan_raw.data.features.copy()
print(df_check.columns.tolist())
print(df_check.filter(regex='(?i)sex').head())
```

Folktables ACS Employment

```python
data_source = ACSDataSource(survey_year='2018', horizon='1-Year', survey='person')
acs_data = data_source.get_data(states=["CA"], download=True)

features, label, group = ACSEmployment.df_to_pandas(acs_data)

print("=== FOLKTABLES ACS EMPLOYMENT ===")
print("Features shape:", features.shape)
print("Columns:", list(features.columns))
group.head()
```

```python
# Folktables, check SEX column's actual values
from folktables import ACSDataSource, ACSEmployment
data_source = ACSDataSource(survey_year='2018', horizon='1-Year', survey='person')
acs_data = data_source.get_data(states=["CA"], download=True)
features, label, group = ACSEmployment.df_to_pandas(acs_data)
print(features['SEX'].value_counts())
print(label.value_counts())
```
